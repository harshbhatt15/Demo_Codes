#### Problem - Perceptron non-linear data pe kaam nhi kr pata

#### Problem - Perceptron does not work with non-Linear data

##### esentially perceptron is a linear model 

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
or_data = pd.DataFrame()
and_data = pd.DataFrame()
xor_data = pd.DataFrame()
```

#### OR Table

```python
or_data['input1'] = [1 , 1 , 0 , 0]
or_data['input2'] = [1 , 0 , 1 , 0]
or_data['output'] = [1 , 1 , 1 , 0]
```

```python
or_data
```

```python
sns.scatterplot(
    data=or_data,
    x='input1',
    y='input2',
    hue='output',
    s=200
)
```

#### AND Table

```python
and_data['input1'] = [1 , 1 , 0 , 0]
and_data['input2'] = [1 , 0 , 1 , 0]
and_data['output'] = [1 , 0 , 0 , 0]
```

```python
and_data
```

```python
sns.scatterplot(
    data=and_data,
    x='input1',
    y='input2',
    hue='output',
    s=250,
    palette='bright'
)
```

#### XOR Table

```python
xor_data['input1'] = [1 , 1 , 0 , 0]
xor_data['input2'] = [1 , 0 , 1 , 0]
xor_data['output'] = [0 , 1 , 1 , 0]
```

```python
xor_data
```

```python
sns.scatterplot(
    data=xor_data,
    x='input1',
    y='input2',
    hue='output',
    s=250,
    palette='bright'
)
```

```python
from sklearn.linear_model import Perceptron
cl1 = Perceptron()
cl2 = Perceptron()
cl3 = Perceptron()
```

```python
cl1.fit(and_data.iloc[:,0:2].values,and_data.iloc[:,-1].values)
cl2.fit(or_data.iloc[:,0:2].values,or_data.iloc[:,-1].values)
cl3.fit(xor_data.iloc[:,0:2].values,xor_data.iloc[:,-1].values)
```

```python
cl1.coef_ , cl1.intercept_
```

```python
x1 = np.linspace(-1,1,5)
y1 = x1+0.5
```

```python
plt.plot(x1,y1)
sns.scatterplot(
    data=or_data,
    x='input1',
    y='input2',
    hue='output',
    s=200
)
```

```python
cl2.coef_ , cl2.intercept_
```

```python
x2=np.linspace(-1,1,5)
y2=-x1+0.5
```

```python
plt.plot(x2,y2)
sns.scatterplot(
    data=or_data,
    x='input1',
    y='input2',
    hue='output',
    s=200
)   
```

```python
cl3.coef_ , cl3.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(xor_data.iloc[:,0:2].values,xor_data.iloc[:,-1].values, clf=cl3, legend=2)
```
